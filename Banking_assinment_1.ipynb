```python
#1.Display all transactions. 2.Print schema, columns and data types.
from pyspark.sql import SparkSession
from pyspark.sql.functions import *

spark = SparkSession.builder.appName("BankingAssignment").getOrCreate()

Banking_data = [
    (1001, 101, "Savings", "Deposit", 5000.00, "Bangalore", "2026-09-01"),
    (1002, 102, "Current", "Withdrawal", 1500.00, "Mumbai", "2026-09-02"),
    (1003, 103, "Savings", "Transfer", 2500.00, "Hyderabad", "2026-09-03"),
    (1004, 104, "Savings", "Deposit", 10000.00, "Chennai", "2026-09-04"),
    (1005, 105, "Current", "Withdrawal", 3200.00, "Delhi", "2026-09-05"),
    (1006, 101, "Savings", "Withdrawal", 800.00, "Bangalore", "2026-09-06"),
    (1007, 106, "Savings", "Deposit", 7500.00, "Pune", "2026-09-07"),
    (1008, 107, "Current", "Transfer", 4500.00, "Kolkata", "2026-09-08"),
    (1009, 108, "Savings", "Deposit", 12000.00, "Bangalore", "2026-09-09"),
    (1010, 109, "Current", "Withdrawal", 2100.00, "Mumbai", "2026-09-10"),
    (1011, 110, "Savings", "Transfer", 3500.00, "Hyderabad", "2026-09-11"),
    (1012, 102, "Current", "Deposit", 6000.00, "Mumbai", "2026-09-12"),
    (1013, 103, "Savings", "Withdrawal", 1800.00, "Hyderabad", "2026-09-13"),
    (1014, 104, "Savings", "Transfer", 9000.00, "Chennai", "2026-09-14"),
    (1015, 105, "Current", "Deposit", 15000.00, "Delhi", "2026-09-15")
]
Cols=["TransactionID","Customer_id","AccountType","TransactionType","Amount","City","TransactionDate"]
banking_df=spark.createDataFrame(Banking_data, Cols)
banking_df.show(truncate=False)
banking_df.printSchema()
print("The columns are " + str(banking_df.columns))
print("Data types are"+str(banking_df.dtypes))
```

```python
#3.Count total transactions.


print("The number of rows is " + str(banking_df.count()))
banking_df.count()
```

```python
#4.Display the first transaction.


print("first transaction is"+str(banking_df.first()))
#banking_df.first()
#display(banking_df.first())
```

```python
#5.Retrieve the first 5 transactions using head

print("first 5 transactions are"+str(banking_df.head(5)))
#banking_df.head(5)
#display(banking_df.head(5)
```

```python
#6.Retrieve 3 rows using take
for rows in banking_df.take(3):
    print(rows)
```

```python
#7.Use collect on a small selection of customer IDs.

banking_df.select("Customer_id").collect()
```

```python
#8.Use countByValue to count transaction types
banking_df.groupBy("TransactionType").count().show()
```

```python
#9.Use reduce to calculate total transaction amount
from pyspark.sql.functions import col, sum

banking_df = banking_df.withColumn("amount_float", col("Amount").cast("float"))
banking_df.agg(sum("amount_float").alias("total_transaction_amount")).show()
```

```python
#10.Check whether the DataFrame is empty.

print("Is this dataframe empty ?"+str(banking_df.isEmpty()))
#banking_df.isEmpty()
```

```python
#11.Iterate through transaction ID and amount using toLocalIterator
for rows in banking_df.select("TransactionID","Amount").toLocalIterator():
    print(rows["TransactionID"],rows["Amount"])
```

```python
#12.Write the banking DataFrame to Parquet.

banking_df.write.mode("overwrite").parquet("/Volumes/workspace/default/banking_data/Banking_data")
df=spark.read.parquet("/Volumes/workspace/default/banking_data/Banking_data")
df.show()
```

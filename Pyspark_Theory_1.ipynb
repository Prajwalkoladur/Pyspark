1. Difference between map() and flatMap() in PySpark

Ans: map() transforms each input element into exactly one output element (one-to-one)

flatMap() transforms each input element into zero or more output elements and flattens the result.

Example: rdd. map (lambda x: x.split()) gives lists rdd.
flatMap (lambda x: x.split()) gives individual words

2.How does the collect() function work in RDD PySpark?

Ans: collect() brings all elements of an RDD from worker nodes to the driver as a Python list. It is an action and should be used only when the result is small because it can cause driver memory overflow

Syntax: rdd.collect()

3) Difference between reduce By Key() and group By Key()

Ans: reduce Bykey() combines values for each key using a reduction function and performs local aggregation before shuffle, so it is generally faster and more memory efficient.

groupBykey() groups all values for each key and shuffles all values, so it can be slower and use more memory.

5) Explain infer Schema in CSV file format. How does it work? Advantages and disadvantages.

Ans: infer Schema tells Spark to automatically detect column data types while reading CSV instead of treating every column as string.
Example: spark read. option ("header", "true"). option ("infer Schema", "true").csv (path).

Advantages: 1) less manual schema definition

 2) detects numeric/date-like types

 3) convenient for exploration.

Disadvantages:
 1) extra time to scan data

 2) possible incorrect type inference

9) Difference between file formats CSV, Avro, Parquet, ORC

Ans: CSV plain text, human-readable, larger and slower, weak schema.

Avro row-oriented, compact, schema-based and good for data exchange/streaming.

Parquet: columnar, compressed, efficient for analytics and widely used in Spark.

ORC: columnar, highly compressed and optimized for Hive/Spark analytical workloads.

Ans: Partitioning divides data into separate directory paths based on column values and is useful for partition pruning.

Partition count depends on distinct partition values

Bucketing distributes data into a fixed number of files/buckets using a hash of bucket columns and can improve joins and aggregations.

bucket count is specified

11) Difference between Spark SQL and DataFrames

Ans: Spark SQL uses SQL queries such as SELECT, WHERE and GROUP BY.

DataFrame API uses programmatic functions such as select(), filter(), groupBy().

Both use Spark's Catalyst optimizer and Tungsten execution engine and generally have similar performance.
